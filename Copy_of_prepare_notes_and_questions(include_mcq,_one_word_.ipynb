## Part 1: Comprehensive Revision Notes

### 1. Electrolytes and Non-Electrolytes

* **Electrolyte:** A chemical compound that conducts an electric current either in its fused (molten) state or in aqueous solution, undergoing chemical decomposition.
* *Examples:* Dilute $\text{H}_2\text{SO}_4$, dilute $\text{HCl}$, molten $\text{PbBr}_2$, aqueous $\text{CuSO}_4$, molten $\text{NaCl}$.


* **Non-Electrolyte:** A chemical compound which neither in solution nor in the molten state conducts electricity and remains undecomposed.
* *Examples:* Distilled water, alcohol, kerosene, sugar solution, urea solution, liquid carbon tetrachloride ($\text{CCl}_4$).


* **Difference in Conduction Mechanism:**
* **Metallic Conduction:** Flow of electric current is due to the movement of **free delocalized electrons**; no chemical change occurs.
* **Electrolytic Conduction:** Flow of electric current is due to the migration of **ions**; accompanied by chemical decomposition.



---

### 2. Classification by Ionic/Molecular Composition

| Category | Particles Present | Degree of Ionisation / Dissociation | Electrical Conductivity | Examples |
| --- | --- | --- | --- | --- |
| **Strong Electrolytes** | **Ions only** (almost completely dissociated) | High ($\approx 100\%$) | High conductor of electricity | Molten $\text{NaCl}$, molten $\text{PbBr}_2$, dil. $\text{HCl}$, dil. $\text{H}_2\text{SO}_4$, $\text{NaOH}$ soln, $\text{CuSO}_4$ soln |
| **Weak Electrolytes** | **Both ions and molecules** | Low (partial ionization) | Poor / weak conductor of electricity | Acetic acid ($\text{CH}_3\text{COOH}$), Carbonic acid ($\text{H}_2\text{CO}_3$), Ammonium hydroxide ($\text{NH}_4\text{OH}$) |
| **Non-Electrolytes** | **Molecules only** | Zero ($0\%$) | Non-conductor | Distilled water, pure ethanol, glucose solution, liquid $\text{CCl}_4$ |

### 3. Basic Terminology & Electronic Concept of Redox

* **Electrolysis:** The process of decomposition of an electrovalent compound in the molten state or aqueous solution by the passage of direct electric current.
* **Electrodes:** Metal plates or carbon rods through which electric current enters or leaves the electrolyte.
* **Anode:** Positive electrode connected to the positive terminal of the DC source; electrons leave the electrolyte here.
* **Cathode:** Negative electrode connected to the negative terminal of the DC source; electrons enter the electrolyte here.


* **Ions:**
* **Cations:** Positively charged ions that migrate to the **cathode**.
* **Anions:** Negatively charged ions that migrate to the **anode**.


* **Electronic Redox Definition:**
* **Oxidation:** Process involving **loss of electrons** (takes place at the **anode**):
$$\text{A}^{n-} - n\text{e}^- \longrightarrow \text{A} \quad \text{or} \quad \text{M} - n\text{e}^- \longrightarrow \text{M}^{n+}$$


* **Reduction:** Process involving **gain of electrons** (takes place at the **cathode**):
$$\text{M}^{n+} + n\text{e}^- \longrightarrow \text{M}$$


* *Conclusion:* Electrolysis is an overall **redox reaction**.



---

### 4. Selective Discharge of Ions & Electrolytic Examples

When multiple cations or anions migrate toward an electrode, preferential discharge is governed by:

1. **Position in Electrochemical Series:** Ions lower in the activity series discharge preferentially because they accept/lose electrons more readily.
* *Cations (ease of discharge increases downward):* $\text{K}^+ < \text{Na}^+ < \text{Ca}^{2+} < \text{Mg}^{2+} < \text{Al}^{3+} < \text{Zn}^{2+} < \text{Fe}^{2+} < \text{Pb}^{2+} < \text{H}^+ < \text{Cu}^{2+} < \text{Ag}^+$
* *Anions (ease of discharge increases downward):* $\text{SO}_4^{2-} < \text{NO}_3^- < \text{Cl}^- < \text{Br}^- < \text{I}^- < \text{OH}^-$


2. **Relative Concentration of Ions:** Higher concentration can override electrochemical position.
3. **Nature of the Electrode:** Inert electrodes (Pt, graphite) vs active electrodes (Cu, Ni, Ag). Active anodes participate in the reaction and dissolve.

### 5. Industrial Applications of Electrolysis

#### A. Electroplating

* **Objective:** To prevent corrosion and enhance aesthetic appearance.
* **Mandatory Conditions:**
* Article to be plated is always made the **Cathode** (cations deposit here).
* Plating metal is made the **Anode** (dissolves to replenish ions).
* Electrolyte must contain a soluble salt of the plating metal.
* Current used must be **Direct Current (DC)** of **low strength for a longer duration** (ensures a uniform, smooth, and firm coating).



```
+---------------------------------------------------------------------------------------------------------------------------+
| Application       | Electrolyte                             | Cathode (Article)       | Anode (Plating metal)             |
+---------------------------------------------------------------------------------------------------------------------------+
| Silver Plating    | Sodium argentocyanide                   | Clean spoon/article     | Block of pure Silver (Ag)         |
|                   | Na[Ag(CN)2] + dil. NaCN                 | Ag+ + e- -> Ag          | Ag - e- -> Ag+                    |
|                   | (Note: AgNO3 is avoided to prevent      |                         |                                   |
|                   | rapid, non-uniform black deposition)    |                         |                                   |
+---------------------------------------------------------------------------------------------------------------------------+
| Nickel Plating    | Aqueous Nickel sulphate [NiSO4]         | Clean iron/brass article| Plate of pure Nickel (Ni)         |
|                   | + small amt of boric acid               | Ni2+ + 2e- -> Ni        | Ni - 2e- -> Ni2+                  |
+---------------------------------------------------------------------------------------------------------------------------+

```

#### B. Electro-refining of Copper

* **Cathode:** Thin strip of pure copper.
* *Reaction:* $\text{Cu}^{2+} + 2\text{e}^- \longrightarrow \text{Cu}$


* **Anode:** Thick slab of impure blister copper.
* *Reaction:* $\text{Cu} - 2\text{e}^- \longrightarrow \text{Cu}^{2+}$


* **Electrolyte:** Aqueous copper(II) sulphate solution acidified with dilute $\text{H}_2\text{SO}_4$.
* **Observation:** Pure copper deposits on the cathode; the anode dissolves away. Impurities like Ag and Au settle beneath the anode as **anode mud / anode slime**.

### Section D: Short Answer Questions

1. **Distinguish between electrolytic dissociation and ionisation with one example each.**
*Answer:*
* **Electrolytic Dissociation:** The separation of pre-existing ions in electrovalent compounds upon melting or dissolution in water.
$$\text{NaCl} \xrightarrow{\text{water}} \text{Na}^+ + \text{Cl}^-$$


* **Ionisation:** The process whereby polar covalent molecules break into oppositely charged ions in aqueous solution.
$$\text{HCl} + \text{H}_2\text{O} \longrightarrow \text{H}_3\text{O}^+ + \text{Cl}^-$$




2. **Why is a silica crucible chosen during the electrolysis of molten lead bromide?**
*Answer:* Silica is non-reactive, heat-resistant, withstands high fusion temperatures without cracking, and is non-conducting. Lead bromide vapors do not corrode silica.
3. **Between $\text{OH}^-$ and $\text{SO}_4^{2-}$ ions, which ion discharges preferentially at the anode and why?**
*Answer:* $\text{OH}^-$ discharges preferentially. In the electrochemical series of anions, $\text{OH}^-$ is positioned lower than $\text{SO}_4^{2-}$ and has a lower discharge potential, releasing electrons with greater ease.

### Section E: Long Answer Questions

1. **Explain the process of electroplating an iron spoon with silver.**
Include: (i) Setup & electrodes, (ii) Electrolyte chosen with reasoning, (iii) Electrode reactions.
*Answer:*
* **Setup:**
* **Cathode:** The iron spoon (thoroughly cleaned with alkali and acid to remove grease/oxide film).
* **Anode:** A plate of pure silver.


* **Electrolyte:** Sodium argentocyanide solution, $\text{Na}[\text{Ag}(\text{CN})_2]$.
* *Reason:* Using $\text{AgNO}_3$ results in very fast deposition, producing an uneven and brittle black layer. $\text{Na}[\text{Ag}(\text{CN})_2]$ provides a controlled, slow supply of $\text{Ag}^+$ ions, creating an adherent, uniform, and lustrous plate.


* **Reactions:**
* **Cathode:** $\text{Ag}^+ + \text{e}^- \longrightarrow \text{Ag}$ (deposited onto spoon)
* **Anode:** $\text{Ag} - \text{e}^- \longrightarrow \text{Ag}^+$ (silver plate dissolves)




2. **Contrast the electrolysis of aqueous $\text{CuSO}_4$ using:**
(a) Platinum electrodes
(b) Copper electrodes
*Answer:*
* **Using Platinum Electrodes (Inert):**
* Cathode: $\text{Cu}^{2+} + 2\text{e}^- \longrightarrow \text{Cu}$ (pink Cu metal deposits)
* Anode: $4\text{OH}^- - 4\text{e}^- \longrightarrow 2\text{H}_2\text{O} + \text{O}_2\uparrow$ (oxygen gas bubbles off)
* Color change: Blue color gradually fades and turns colorless as $\text{Cu}^{2+}$ ions are consumed without being replaced.


* **Using Copper Electrodes (Active):**
* Cathode: $\text{Cu}^{2+} + 2\text{e}^- \longrightarrow \text{Cu}$ (deposits copper, cathode gains weight)
* Anode: $\text{Cu} - 2\text{e}^- \longrightarrow \text{Cu}^{2+}$ (copper anode dissolves, loses weight)
* Color change: Blue color remains constant because the rate of formation of $\text{Cu}^{2+}$ ions at the anode equals their discharge rate at the cathode.

## Part 3: ICSE Previous 10 Years Board Exam Questions & Solutions

### ICSE 2024

* **Q1:** State the reaction occurring at the cathode during electrorefining of copper.
*Solution:* $\text{Cu}^{2+} + 2\text{e}^- \longrightarrow \text{Cu}$
* **Q2:** Why does liquefied pure hydrogen chloride not conduct electricity, while its aqueous solution does?
*Solution:* Pure liquefied $\text{HCl}$ consists only of neutral covalent molecules (no free ions). In water, the polar water molecules cause $\text{HCl}$ to ionise into free $\text{H}_3\text{O}^+$ and $\text{Cl}^-$ ions, which conduct electricity.

### ICSE 2023

* **Q:** Name the gas released at the anode during the electrolysis of acidified water using platinum electrodes.
*Solution:* Oxygen gas ($\text{O}_2$). Reaction: $4\text{OH}^- - 4\text{e}^- \longrightarrow 2\text{H}_2\text{O} + \text{O}_2\uparrow$.

### ICSE 2022 (Semester 1 / 2)

* **Q:** Identify the weak electrolyte among:
(a) Dil. $\text{H}_2\text{SO}_4$, (b) Dil. $\text{HCl}$, (c) $\text{NH}_4\text{OH}$, (d) Molten $\text{NaCl}$
*Solution:* **(c) $\text{NH}_4\text{OH}$** (it undergoes partial ionisation and contains both ions and molecules).

### ICSE 2021 (Special / Model Review)

* **Q:** What is observed at the cathode and anode when molten lead bromide is electrolysed?
*Solution:*
* Cathode: Silvery-grey lead metal is deposited.
* Anode: Reddish-brown vapors of bromine gas are evolved.



### ICSE 2020

* **Q:** Write the ionic equation for the reaction occurring at the anode during silver plating of an iron spoon.
*Solution:* $\text{Ag} - \text{e}^- \longrightarrow \text{Ag}^+$ (or $\text{Ag} \longrightarrow \text{Ag}^+ + \text{e}^-$)

### ICSE 2019

* **Q:** During the electrolysis of copper(II) sulphate using platinum electrodes, an anode reaction takes place. Write the anode equation and name the gas evolved.
*Solution:* Anode equation: $4\text{OH}^- - 4\text{e}^- \longrightarrow 2\text{H}_2\text{O} + \text{O}_2\uparrow$. Gas evolved: **Oxygen gas**.

### ICSE 2018

* **Q1:** The electrolysis of acidified water is an example of: (a) reduction, (b) oxidation, (c) redox reaction, (d) synthesis.
*Solution:* **(c) redox reaction**
* **Q2:** Why are graphite anodes preferred over other inert electrodes in the electrolysis of molten $\text{PbBr}_2$?
*Solution:* Graphite is unaffected by reactive, corrosive bromine vapors at high temperatures.

### ICSE 2017

* **Q:** Differentiate between electrical conductivity of copper wire and copper sulphate solution.
*Solution:*
* **Copper wire:** Conducts electricity through mobile **electrons**; undergoes **no chemical decomposition**.
* **Copper sulphate solution:** Conducts through mobile **ions** ($\text{Cu}^{2+}$, $\text{SO}_4^{2-}$); undergoes **chemical decomposition**.



### ICSE 2016

* **Q:** The particles present in a strong electrolyte are:
(a) only molecules, (b) mainly ions, (c) ions and molecules, (d) only atoms.
*Solution:* **(b) mainly ions**

### ICSE 2015

* **Q:** Select the correct choice from the brackets:
"During electroplating of an article with nickel, the electrolyte used is _______ (acidified $\text{NiSO}_4$ / $\text{Ni(NO}_3\text{)}_2$ / $\text{NiCl}_2$)."
*Solution:* **Aqueous acidified nickel sulphate ($\text{NiSO}_4$)**.
